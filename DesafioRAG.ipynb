```python

# Vamos usar o texto a partir da página 34 do PDF
pagina_inicio = 34

paginas_livro = [
    pagina for pagina in paginas
    if pagina["pagina"] >= pagina_inicio
]

print("Páginas selecionadas:", len(paginas_livro))
print("\nPrimeiro trecho selecionado:\n")
print(paginas_livro[0]["texto"][:1000])
```

```python

def dividir_texto(texto, tamanho=1000, sobreposicao=200):
    chunks = []
    inicio = 0

    while inicio < len(texto):
        fim = inicio + tamanho
        trecho = texto[inicio:fim].strip()

        if trecho:
            chunks.append(trecho)

        inicio += tamanho - sobreposicao

    return chunks


chunks = []

for pagina in paginas_livro:
    trechos = dividir_texto(pagina["texto"])

    for trecho in trechos:
        chunks.append({
            "texto": trecho,
            "pagina": pagina["pagina"]
        })

print("Total de chunks:", len(chunks))
print("\nExemplo do primeiro chunk:\n")
print(chunks[0]["texto"])
print("\nPágina do PDF:", chunks[0]["pagina"])
```

```python

from sentence_transformers import SentenceTransformer

# Modelo multilíngue que funciona com textos em português
modelo_embeddings = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

# Extrair o texto de cada chunk
textos = [chunk["texto"] for chunk in chunks]

# Transformar os textos em vetores numéricos
embeddings = modelo_embeddings.encode(
    textos,
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=True
)

# Garantir o formato esperado pelo FAISS
embeddings = embeddings.astype("float32")

print("Formato dos embeddings:", embeddings.shape)
print("Quantidade de chunks:", len(chunks))
```

```python

print("Modelo carregado:", modelo_embeddings is not None)
print("Chunks:", len(chunks))
print("Embeddings criados:", embeddings.shape)
print("Tipo dos dados:", embeddings.dtype)
```

```python
import faiss

dimensao = embeddings.shape[1]

indice = faiss.IndexFlatIP(dimensao)
indice.add(embeddings)

print("Total de vetores no índice:", indice.ntotal)
```

```python
def buscar_trechos(pergunta, quantidade=4):
    vetor_pergunta = modelo_embeddings.encode(
        [pergunta],
        convert_to_numpy=True,
        normalize_embeddings=True
    ).astype("float32")

    pontuacoes, indices_encontrados = indice.search(
        vetor_pergunta, quantidade
    )

    resultados = []

    for pontuacao, idx in zip(
        pontuacoes[0], indices_encontrados[0]
    ):
        if idx == -1:
            continue

        resultados.append({
            "texto": chunks[idx]["texto"],
            "pagina": chunks[idx]["pagina"],
            "similaridade": float(pontuacao)
        })

    return resultados
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

nome_modelo = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizador = AutoTokenizer.from_pretrained(nome_modelo)

modelo_gerador = AutoModelForCausalLM.from_pretrained(
    nome_modelo,
    torch_dtype=torch.float32
)

print("Modelo carregado com sucesso!")
```

```python
def gerar_resposta(pergunta, contexto):
    prompt = montar_prompt(pergunta, contexto)

    mensagens = [
        {
            "role": "user",
            "content": prompt
        }
    ]

    texto_formatado = tokenizador.apply_chat_template(
        mensagens,
        tokenize=False,
        add_generation_prompt=True
    )

    entradas = tokenizador(
        texto_formatado,
        return_tensors="pt"
    )

    with torch.no_grad():
        saida = modelo_gerador.generate(
            **entradas,
            max_new_tokens=250,
            do_sample=False,
            pad_token_id=tokenizador.eos_token_id
        )

    tokens_novos = saida[0][entradas["input_ids"].shape[1]:]

    resposta = tokenizador.decode(
        tokens_novos,
        skip_special_tokens=True
    )

    return resposta.strip()
```

```python
from collections import defaultdict

textos_por_pagina = defaultdict(list)

for chunk in chunks:
    textos_por_pagina[chunk["pagina"]].append(chunk["texto"])

print("Páginas com trechos:", len(textos_por_pagina))
print("Exemplo de página:", next(iter(textos_por_pagina)))
```

```python
tamanho_pai = 5

paginas_ordenadas = sorted(textos_por_pagina.keys())

trechos_pai = []

for inicio in range(0, len(paginas_ordenadas), tamanho_pai):
    grupo_paginas = paginas_ordenadas[inicio:inicio + tamanho_pai]

    texto_pai = "\n".join(
        " ".join(textos_por_pagina[pagina])
        for pagina in grupo_paginas
    )

    trechos_pai.append({
        "paginas": grupo_paginas,
        "texto": texto_pai
    })

print("Total de trechos-pai:", len(trechos_pai))
print("\nExemplo de trecho-pai:")
print("Páginas:", trechos_pai[0]["paginas"])
print(trechos_pai[0]["texto"][:1000])
```

```python
# Relaciona cada página ao índice do trecho-pai correspondente
pagina_para_pai = {}

for indice_pai, pai in enumerate(trechos_pai):
    for pagina in pai["paginas"]:
        pagina_para_pai[pagina] = indice_pai

print("Páginas relacionadas:", len(pagina_para_pai))
print("Página 34 pertence ao trecho-pai:", pagina_para_pai[34])
print("Página 83 pertence ao trecho-pai:", pagina_para_pai[83])
```

```python
def buscar_parent_rag(pergunta, quantidade=4):
    # Busca nos chunks pequenos já indexados
    resultados_pequenos = buscar_trechos(
        pergunta,
        quantidade=quantidade
    )

    pais_encontrados = []
    indices_pais_vistos = set()

    for resultado in resultados_pequenos:
        pagina = resultado["pagina"]
        indice_pai = pagina_para_pai[pagina]

        # Evita devolver o mesmo trecho-pai mais de uma vez
        if indice_pai not in indices_pais_vistos:
            indices_pais_vistos.add(indice_pai)

            pai = trechos_pai[indice_pai]

            pais_encontrados.append({
                "paginas": pai["paginas"],
                "texto": pai["texto"],
                "similaridade": resultado["similaridade"]
            })

    return pais_encontrados
```

```python
resultados_parent = buscar_parent_rag(
    "Como a seca e as condições naturais do sertão afetam a vida dos sertanejos?",
    quantidade=4
)

for i, resultado in enumerate(resultados_parent, 1):
    print(
        f"\n--- Resultado {i} | "
        f"Páginas PDF {resultado['paginas'][0]}–"
        f"{resultado['paginas'][-1]} ---"
    )
    print(resultado["texto"][:1000])
```

```python
from sentence_transformers import CrossEncoder

modelo_reranker = CrossEncoder(
    "cross-encoder/mmarco-mMiniLMv2-L12-H384-v1",
    max_length=512
)

print("Modelo de reranking carregado!")
```

```python
def buscar_rerank_rag(pergunta, quantidade_inicial=12, quantidade_final=4):
    # 1. Busca inicial usando FAISS
    candidatos = buscar_trechos(
        pergunta,
        quantidade=quantidade_inicial
    )

    if not candidatos:
        return []

    # 2. Compara a pergunta com cada trecho
    pares = [
        (pergunta, candidato["texto"])
        for candidato in candidatos
    ]

    pontuacoes = modelo_reranker.predict(
        pares,
        show_progress_bar=False
    )

    # 3. Ordena pelos resultados do reranker
    for candidato, pontuacao in zip(candidatos, pontuacoes):
        candidato["pontuacao_reranker"] = float(pontuacao)

    candidatos_ordenados = sorted(
        candidatos,
        key=lambda x: x["pontuacao_reranker"],
        reverse=True
    )

    return candidatos_ordenados[:quantidade_final]
```

```python
resultados_rerank = buscar_rerank_rag(
    "Como a seca e as condições naturais do sertão afetam a vida dos sertanejos?"
)

for i, resultado in enumerate(resultados_rerank, 1):
    print(
        f"\n--- Resultado {i} | "
        f"Página PDF {resultado['pagina']} | "
        f"Pontuação {resultado['pontuacao_reranker']:.3f} ---"
    )
    print(resultado["texto"][:700])
```
